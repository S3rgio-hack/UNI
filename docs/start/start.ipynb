```
# Celda 2: Simulación de un Qubit
import cirq
import numpy as np

# 1. Definir un qubit en una cuadrícula o línea
qubit = cirq.GridQubit(0, 0)

# 2. Crear un circuito: Compuerta H (Hadamard) + Medición
# La compuerta H transforma |0> en (|0> + |1>) / sqrt(2)
circuito_1q = cirq.Circuit(
    cirq.H(qubit),
    cirq.measure(qubit, key='resultado')
)

print("--- Circuito de 1 Qubit ---")
print(circuito_1q)

# 3. Simular el comportamiento con cirq.Simulator
simulador = cirq.Simulator()
resultado_1q = simulador.run(circuito_1q, repetitions=1000)

# 4. Analizar frecuencias
frecuencias_1q = resultado_1q.histogram(key='resultado')
print("\nFrecuencias de medición (1000 corridas):")
print(frecuencias_1q)
```

```
# Celda 3: Generación de un Estado de Bell
# Definir dos qubits
q0, q1 = cirq.LineQubit.range(2)

# Crear circuito de entrelazamiento (Bell)
circuito_bell = cirq.Circuit(
    cirq.H(q0),          # Pone al q0 en superposición
    cirq.CNOT(q0, q1),   # Entrelaza q0 con q1
    cirq.measure(q0, q1, key='bell_result')
)

print("--- Circuito Estado de Bell ---")
print(circuito_bell)
```

```
# Celda 4: Simulación y análisis de frecuencias del Estado de Bell
resultado_bell = simulador.run(circuito_bell, repetitions=1000)
histograma_bell = resultado_bell.histogram(key='bell_result')

print("Resultados de las mediciones (Estado de Bell):")
print(histograma_bell)
```

```
# Celda 5: Comparación entre cirq.Simulator y qsimcirq.QSimSimulator
import time
import qsimcirq

# 1. Simulación con cirq.Simulator (Estado puro en Python)
inicio_cirq = time.time()
sim_estandar = cirq.Simulator()
res_cirq = sim_estandar.run(circuito_bell, repetitions=5000)
tiempo_cirq = time.time() - inicio_cirq

# 2. Simulación con qsimcirq.QSimSimulator (Backend de alto rendimiento optimizado)
inicio_qsim = time.time()
sim_qsim = qsimcirq.QSimSimulator()
res_qsim = sim_qsim.run(circuito_bell, repetitions=5000)
tiempo_qsim = time.time() - inicio_qsim

print(f"Tiempo con cirq.Simulator: {tiempo_cirq:.6f} segundos")
print(f"Tiempo con qsimcirq.QSimSimulator: {tiempo_qsim:.6f} segundos")

print("\nHistograma con qsim:")
print(res_qsim.histogram(key='bell_result'))
```
